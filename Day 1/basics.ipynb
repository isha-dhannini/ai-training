```python
import numpy as np

# 1D Array
a = np.array([1, 2, 3, 4, 5])

print('Array:', a)
print('Shape:', a.shape)
print('Size:', a.size)
print('Data type:', a.dtype)
print('Number of dimensions:', a.ndim)
```

```python
# 2D Array (Matrix)
mat = np.array([[1, 2, 3], [4, 5, 6]])

print('\nArray:\n', mat)
print('Shape:', mat.shape)
print('Size:', mat.size)
print('Data type:', mat.dtype)
print('Number of dimensions:', mat.ndim)
```

```python
# Convenience Constructors

zeros = np.zeros((2, 3))
print('\nZeros array\n', zeros)

ones = np.ones((3, 4))
print('\nOnes array\n', ones)

eye = np.eye(4)
print('\nIdentity matrix\n', eye)

arr_range = np.arange(0, 10, 2)
print('\nArange array\n', arr_range)

linspace = np.linspace(0, 1, 5)
print('\nLinspace array\n', linspace)
```
